### Functies laden

De map `open_loop` is het Python-package. Het bestand `meting.py` daarin bevat de functie `voer_open_loop_meting_uit`. Met onderstaande import laden we die functie.

```python
from open_loop.meting import voer_open_loop_meting_uit
import matplotlib.pyplot as plt
```

### Instellingen en meting

Pas de waarden aan en voer de cel uit om de motor te starten. De motor stabiliseert eerst 2 seconden en meet daarna gedurende `meetduur` seconden.

```python
com_poort = "COM4"
pwm = 50
meetduur = 10.0     # seconden
delta_t = 0.01    # seconden tussen meetpunten

metingen, csv_pad = voer_open_loop_meting_uit(
    com_poort=com_poort,
    pwm=pwm,
    meetduur=meetduur,
    delta_t=delta_t,
)

print(csv_pad)
```

### Grafiek

```python
import pandas as pd
import numpy as np

data = pd.read_csv("data/aeropendulum_20261112_132049_435893.csv")

tijd = data["tijd_s"].to_numpy()
hoek = data["hoek_deg"].to_numpy()
```
